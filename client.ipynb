```python
import socket

server =socket.socket()
server.bind(("127.0.0.1",5000))
server.listen(1)

print("Waiting for client")
conn,addr=server.accept()
print("Client connected")
```

```python
message=conn.recv(1024).decode()
print("Client:",message)
conn.send("Hello Client".encode())
conn.close()
server.close()
```
